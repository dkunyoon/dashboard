```python
# 각 메서드별 지원 provider 확인
for method in ["price", "reference_rates", "search", "snapshots"]:
    try:
        sub = getattr(obb.currency, method)
        submethods = [x for x in dir(sub) if not x.startswith("_")]
        print(f"currency.{method}: {submethods}")
    except Exception as e:
        print(f"currency.{method}: ERROR - {e}")
```

```python
# fmp 무료 플랜 - 주요 페어만 가능한지 테스트
for symbol in ["EURUSD", "USDKRW", "GBPUSD", "USDJPY", "AUDUSD", "USDCAD", "NZDUSD", "BTCUSD"]:
    try:
        df = obb.currency.price.historical(symbol=symbol, provider="fmp").to_df()
        print(f"✓ {symbol}: {len(df)}행, {df.index.min()} ~ {df.index.max()}")
    except Exception as e:
        print(f"✗ {symbol}: 유료 전용")
```

```python
# FMP 무료 플랜 - 가능한 페어 historical data 일괄 조회
free_pairs = ["EURUSD", "GBPUSD", "USDJPY", "AUDUSD", "USDCAD", "NZDUSD", "BTCUSD"]

df_hist_fmp = obb.currency.price.historical(
    symbol=",".join(free_pairs),
    provider="fmp"
).to_df()

df_hist_fmp
```

```python
# start_date / end_date 지정 가능 (문자열 "YYYY-MM-DD" 형식)
df_hist_fmp2 = obb.currency.price.historical(
    symbol="USDJPY",
    provider="fmp",
    start_date="2016-01-01",
    end_date="2026-05-07"
).to_df()

print(f"{len(df_hist_fmp2)}행, {df_hist_fmp2.index.min()} ~ {df_hist_fmp2.index.max()}")
df_hist_fmp2.head()
```
